# Ejemplo de conexión y consulta a mi base de datos postgres en mi contenedor, usando puramente psql, sin liberías de python. SQL puro

```python
%%bash
psql postgresql://root:root@localhost:5432/postgres -c "SELECT 1;"
```

```python
%%bash
psql postgresql://root:root@localhost:5432/postgres <<'SQL'
SELECT
    table_schema,
    table_name
FROM information_schema.tables
ORDER BY table_schema, table_name
LIMIT 10;
SQL
```

```python
%%sql
SELECT
    table_schema,
    table_name
FROM information_schema.tables
WHERE table_type = 'BASE TABLE'
ORDER BY
    table_schema,
    table_name;
```
